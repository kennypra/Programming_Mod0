## Comparing Floats

Floats are stored approximately, so 2 values that look equal may not be.

```python
print(0.1 + 0.2)          # not exactly 0.3
```

```python
print(0.3)                # the value we expected
```

```python
print((0.1 + 0.2) == 0.3)   # so they are not equal
```

Floats carry tiny rounding errors, so a sum can miss its target by a tiny amount. Do not compare floats with `==`. Instead check they are close enough, for example with `round()`.

## Equal Value, Different Object

`==` checks value. `id()` checks identity. They can disagree.

```python
a = 1
b = 1.0
print(type(a))        # int
print(type(b))        # float
print(a == b)          # True
print(id(a) == id(b))  # False
```

Different types → different identity, even with equal value.

```python
a = [1, 2]
b = [1, 2]
print(a == b)          # True
print(id(a) == id(b))  # False
```

SAME type, SAME value — still two separate objects.

## Lists Change in Place

A list method changes the list in place — same object, so `id()` stays the same. A string method returns a new object — so you must reassign, and `id()` changes.

```python
lst = [1, 2, 3]
lst.append(4)
```

## Lists Change in Place

You can verify this yourself. Run the code below and compare the id and value of the list and the string before and after each operation.

```python
word = "hi"
print(word)
print(id(word))
word = word.upper()  # creates a new string object since strings are immutable
print(word)
print(id(word))
```

```python
lst = [1, 2, 3]
print(lst)
print(id(lst))
lst.append(4)  # append is in-place operation
print(lst)
print(id(lst))
```

## sort(): sorts the list in place

```python
nums = [3, 1, 2]
print(id(nums))
print(nums)
nums.sort()
print(id(nums))
print(nums)
```

`sort()` sorts the list in place and returns `None`, not a sorted copy.

## sort() vs sorted()

```python
nums = [3, 1, 2]
print(id(nums))
nums.sort()     # mutates in place
print(id(nums))
```

```python
nums = [3, 1, 2]
print(id(nums))
result = sorted(nums)
print(id(result))
```

Method vs. function: `sort()` changes the list itself (same id); `sorted()` hands you a new one (new id).

## pop(): removes and returns an item

```python
nums = [10, 20, 30]
x = nums.pop()
print(x)     # removed item
print(nums)  # now shorter
```

```python
last = nums.pop(-1)
print(nums)
print(last)
```

`pop()` returns AND mutates.

```python
nums.pop(5)   # no index 5 -> IndexError
```

```python
[].pop()      # empty list -> IndexError
```

`pop()` breaks the usual rule: most list methods return `None`, but `pop()` returns the item it removes. `pop(i)` removes at index `i`; with no argument it removes the last item.

## Shared Reference

Shared reference means two or multiple variables point to the same object.

```python
a = [1, 2, 3]
b = a
print(id(a) == id(b))  # True, a and b now point to the SAME object
```

```python
a.append(4)
print(a)    # [1, 2, 3, 4]
print(b)    # [1, 2, 3, 4]  -- b changed too!
```

Since `append` changes the list in place, any change made through `a` also shows up in `b`, because they point to the same object.

## List VS String for Shared Reference

```python
a = [1, 2]
b = a
a.append(3)
print(id(a) == id(b))  # True
```

```python
a = 'hi'
b = a
a = a.upper()
print(id(a) == id(b))  # False
```

**Discuss:** Why does the id of the string change after `a = a.upper()`? Is `a` and `b` still pointing to the same object after this step?

a.upper() create a new string object which is then re assigned to a. Thus, the id of a changes

## Copy or Same List?

```python
nums = [1, 2, 3]
copy = nums[:]
copy[0] = 99


print(id(nums))
print(id(copy))
```

Does `nums` change too? **Hint:** check the `id` of `nums` and `copy`. Make a guess, run the code to verify, then try to explain why.

Answer : nums[:] makes list copy of the original list. Thus, they are different objects with different ids

```python
# Try it yourself: check id(nums) and id(copy) here
print(nums)
print(copy)
```

## Copying a List

A copy of a list holds the same values in a brand-new object, so changing one does not affect the other.

**Make a copy: `a.copy()` or `a[:]`**

```python
a = [1, 2, 3]
b = a.copy()   # or  b = a[:]
a.append(4)
print(a)  # a changed
print(b)  # b unchanged
```

**Check with `==` and `is`**

```python
a = [1, 2, 3]
b = a.copy()
print(a == b)  # same values
print(a is b)  # different object
```

## Quick Summary

- **Operator overloading:** `+` means something different by type (numbers, text, lists) — but never mixes types
- **Equality vs. identity:** `==` compares value; `id()` compares identity — equal value does not mean same object
- **Lists:** an ordered, mutable collection — brackets, any type, even mixed
- **len(), list():** same `len()` as strings; `list()` builds a list from anything ordered
- **Slicing:** the exact same `start:stop:step` rule as strings
- **Methods vs. functions:** `append()`/`sort()` mutate and return `None`; `sorted()` returns a new list; `pop()` is the exception
- **Shared reference:** `b = a` means two names, one object
