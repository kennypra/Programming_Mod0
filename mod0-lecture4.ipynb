## Logical Operators
`and` and `or` combine two conditions into one.
`not` flips a True/False value to its opposite.

They usually combine comparisons:

````python
print((5 > 3) and (2 < 4))
````

## `and`
`and` only evaluates to True if both conditions are True. If either condition is False, the result is False.

| Condition 1 | Condition 2 | Result |
|---|---|---|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

## `or`
`or` evaluates to True if at least one condition is True. It is False only when both conditions are False.

| Condition 1 | Condition 2 | Result |
|---|---|---|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |

## `not`
`not` reverses a single condition. It turns True into False, and False into True.

| Condition | Result |
|---|---|
| True | False |
| False | True |

## Operator Precedence: `not` → `and` → `or`
When operators are combined, Python applies them in a fixed order.

1. `not`
2. `and`
3. `or`

`and` has a higher rank than `or`, so `False and False` is grouped first, equally the same as:

## Parentheses
If we want to change the precedence, we can manually add parentheses to group the conditions.

`True or False and False` → `True` (already verified above).

If we want to have `True or False` as a group, simply add parentheses:

````python
print((True or False) and False)
````

## Short Circuit
Python reads conditions left to right and stops as soon as the result is known.

````python
print((2 > 10) and (9 > 1))
````

`(2 > 10)` is False, so `and` never checks the right side (because `and` requests both conditions to be True).

````python
print((6 > 1) or (9 > 20))
````

`(6 > 1)` is True, so `or` never checks the right side.

**Short circuit rule:** `or` stops at the first True. `and` stops at the first False.

## Quick Summary

**Precedence**
- Order: `not` → `and` → `or`
- When a higher order operator is encountered, group it by adding parentheses (note: this step only groups; it does not evaluate yet)
- Parentheses override the order

**Short Circuit**
- Runs left to right, stops once the result is known.
- `or` stops at the first True
- `and` stops at the first False

## Nested Logic
Nested logic puts one logical expression inside another.
Parentheses mark the inner part, which Python evaluates first.

````python
data1 = "clean"
data2 = "raw"
data3 = "validated"

print(not (not data1 == "clean" or data2 == "raw") and data3 != "validated")
````

### Evaluate the inner part first:

````python
print((not data1 == "clean") or data2 == "raw")
````

Operator Precedence: `not` → `and` → `or` → this inner part evaluates to `True`.

`not (True) and data3 != "validated"`

## Human vs Machine in the Sequence of Operations
Python must go left to right. You can scan the whole line and take a shortcut.

```
not (a and (b or not c)) and (d or e) or 10 > 0
```

**Machine**
- Evaluates strictly left to right.
- Must finish the nested left group first.

**You**
- Scan the whole line first.
- `10 > 0` is True
- So skip the messy left part.

## Summary

**Operators**
- `and`: both must be True.
- `or`: at least one True.
- `not`: flips the value.

**Precedence & Parentheses**
- Order: `not` → `and` → `or`.
- Parentheses override it.

**Short Circuit**
- Left to right, stop when known.
- `or`: stop at first True.
- `and`: stop at first False.

**Watch Out**
- Floats: `0.1 + 0.2` is not `0.3`.
- Nested: read it inside out.
- You shortcut, Python cannot.
