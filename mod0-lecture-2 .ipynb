## Problem Decomposition
If we look at the 3 questions together, we have actually been solving one larger problem:

**Obtain a list of all users’ first names.**

We broke it into smaller steps:
1. Normalize → convert all letters to lowercase (Quick check 1)
2. Separate → split the string into a list of emails (Quick check 1)
3. Extract → get the first name from one email (Quick check 2)
4. Repeat → apply the same process to all emails (Quick check 3)

This is **problem decomposition**: breaking a complex problem into smaller, manageable problems.

*We will revisit this concept throughout the course.*
