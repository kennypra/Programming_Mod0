## Introducing the for loop

- A for loop **repeats** a block of code once for each item in a sequence.
- It always processes items **in the order** they appear.
- You use it when you already know **how many times** the loop needs to run.

A string and a list are both sequences — that is why the same for loop syntax worked on both examples above.

## The range() function

- range(stop) generates whole numbers starting at 0, up to (but not including) stop.
- range(start, stop) begins at start instead of 0, and still stops before stop.

```python
for i in range(5):
    print(i)
```

range(5) produces exactly 5 numbers — 0 through 4. This loop runs exactly 5 times.

```python
for i in range(1, 5):
    print(i)
```

range(1, 5) starts at 1 instead of 0, and still stops before 5 — so this loop runs only 4 times.

## Using range() to Loop Through a List

- range(5) ran the loop body exactly 5 times — that 5 came from typing it directly.
- If a list has 5 elements, len(lst) also gives you 5 — so range(len(lst)) runs the loop exactly once per element, no matter how long the list is.

```python
lst = [20, 10, 40, 60, 50]
len(lst)              # 5
range(len(lst))       # range(0, 5) -> 0, 1, 2, 3, 4
```

lst has 5 elements, so range(len(lst)) is exactly the range(5) you just saw — just computed automatically from the list's length instead of typed by hand.

## range() vs. list(range())

range() on its own is a lazy sequence — Python does not build the full list of numbers until something asks for it. Wrapping it in list() forces Python to generate every value right now.

```python
numbers = list(range(5))
print(numbers)
```

This is the same trick you already used with list("abc") — list() materializes whatever sequence you give it into an actual list you can print, index, or loop over directly.

## Two ways to iterate

Both loops below visit every element of lst, once each — just reached differently.

```python
lst = [20, 10, 40, 60, 50]
for element in lst:
    print(element)
```

lst has 5 elements. Each pass, Python just does a normal assignment: element = lst[i]. element is a regular variable — it just gets reassigned automatically, once per item. There is no index at all in this style.

```python
lst = [20, 10, 40, 60, 50]
for i in range(len(lst)):
    print(lst[i])
```

len(lst) is 5, so range(len(lst)) produces 0, 1, 2, 3, 4 — never 5. i is just an int counting 0 up to 4. lst[i] does the actual look-up into the list at that position. Same repetition as before — the body ran once per item — just reached by counting an index instead of naming the element directly.

## Does the loop variable change the list?

Predict: after each loop below runs, what does lst look like? Then run the cell to check.

```python
lst = [3, 6, 9]
for val in lst:
    val = val * 2
print(lst)
```

val gets REASSIGNED to a new int each pass. That only repoints the name val — the list itself is never indexed into or modified. lst prints unchanged: [3, 6, 9].

```python
lst = [3, 6, 9]
for i in range(len(lst)):
    lst[i] = lst[i] * 2
print(lst)
```

lst[i] = ... is item assignment on the list itself — same id, slot i really changes. Every pass mutated the actual list object at a specific index, so all three changes stuck: [6, 12, 18].

## When the Index Runs Out

Guess first — what happens when this runs? Then run the cell to check.

```python
lst = [10, 20, 30]
for i in range(len(lst) + 1):
    print(lst[i])
```

len(lst) is 3, so range(len(lst) + 1) produces 0, 1, 2, 3 — one index too many. lst only has valid indices 0-2, so asking for lst[3] crashes the loop with IndexError: list index out of range.

Why: the +1 was never needed. range(len(lst)) alone already produces exactly one index per element.

## Changing a List While Looping Over It

Guess first — what does lst look like after this loop finishes? Then run the cell to check.

```python
lst = [1, 2, 2, 3, 4]
for item in lst:
    if item % 2 == 0:
        lst.remove(item)
print(lst)
```

Removing an item shifts every later element one slot to the left. The loop keeps counting up by index regardless — so the element that slides into the just-emptied slot gets skipped entirely. Final result: lst = [1, 2, 3] — the second 2 survives.

Why: a for loop over a list walks by index behind the scenes. Mutating the list's length mid-loop desyncs those indices from the items you actually meant to visit.

## range(len(lst)) Locks In the Length Up Front

Guess first — what happens when this runs? Then run the cell to check.

```python
lst = [1, 2, 2, 3, 4]
for i in range(len(lst)):
    if lst[i] % 2 == 0:
        lst.remove(lst[i])
print(lst)
```

range(len(lst)) is evaluated a single time, before the loop's first pass. Removing items shrinks lst as you go, but the range never shrinks with it — so it eventually asks for an index that no longer exists: IndexError: list index out of range, right when i = 4 and lst only has 3 elements left.
